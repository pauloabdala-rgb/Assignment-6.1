```python
import pandas as pd

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report
```

```python
url = "https://raw.githubusercontent.com/PacktWorkshops/The-Applied-Artificial-Intelligence-Workshop/master/Datasets/car.csv"

df = pd.read_csv(url)

print(df.head())
print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())
```

```python
encoder = LabelEncoder()

for column in df.columns:
    df[column] = encoder.fit_transform(df[column])

print(df.head())
```

```python
label = df.pop("class")
features = df

print("Features:", features.shape)
print("Labels:", label.shape)
```

```python
features_train, features_test, label_train, label_test = train_test_split(
    features,
    label,
    test_size=0.10,
    random_state=42
)

print("Training:", features_train.shape)
print("Testing:", features_test.shape)
```

```python
model = DecisionTreeClassifier(random_state=42)

model.fit(features_train, label_train)

predictions = model.predict(features_test)

print("Predictions:", predictions[:10])
```

```python
accuracy = model.score(features_test, label_test)

print(f"Model Accuracy: {accuracy:.2%}")
```

```python
print(classification_report(
    label_test,
    predictions,
    target_names=["acc", "good", "unacc", "vgood"]
))
```
