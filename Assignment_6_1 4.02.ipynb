```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.metrics import confusion_matrix, classification_report

url = "https://raw.githubusercontent.com/PacktWorkshops/The-Applied-Artificial-Intelligence-Workshop/master/Datasets/car.csv"

df = pd.read_csv(url)
print(df.head())
print(df.shape)
```

```python
encoder = LabelEncoder()

for column in df.columns:
    df[column] = encoder.fit_transform(df[column])

label = df.pop("class")
features = df

features_train, features_test, label_train, label_test = train_test_split(
    features, label, test_size=0.10, random_state=42
)
```

```python
rf_model = RandomForestClassifier(
    n_estimators=100, random_state=42
)

rf_model.fit(features_train, label_train)

rf_predictions = rf_model.predict(features_test)

print("Random Forest Accuracy:", rf_model.score(features_test, label_test))
```

```python
print("Random Forest Confusion Matrix:")
print(confusion_matrix(label_test, rf_predictions))
```

```python
print("Random Forest Classification Report:")
print(classification_report(
    label_test,
    rf_predictions,
    target_names=["acc", "good", "unacc", "vgood"]
))
```

```python
print("Random Forest Feature Importance:")

print(pd.Series(
    rf_model.feature_importances_,
    index=features.columns
).sort_values(ascending=False))
```

```python

extra_model = ExtraTreesClassifier(
    n_estimators=100, random_state=42
)

extra_model.fit(features_train, label_train)

extra_predictions = extra_model.predict(features_test)

print("Extra Trees Accuracy:",
      extra_model.score(features_test, label_test))

print("\nExtra Trees Confusion Matrix:")
print(confusion_matrix(label_test, extra_predictions))

print("\nExtra Trees Classification Report:")
print(classification_report(
    label_test,
    extra_predictions,
    target_names=["acc", "good", "unacc", "vgood"]
))

print("\nExtra Trees Feature Importance:")
print(pd.Series(
    extra_model.feature_importances_,
    index=features.columns
).sort_values(ascending=False))
```

Conclusion:

I applied two classifiers, namely, Random Forest and Extra Trees to Car Evaluation dataset with this activity. The two models were trained with 90% of the data and tested with 10% of the data. I did the comparison based on accuracy, confusion matrices and classification reports. I also wanted to look into feature importance so as to find out what features of a car affected the conclusion. These are the ensembles method that offer useful approaches to improve car classification.
